1. StandardScaler

Avoid When-->Data has many extreme outliers

use when--> Data is approximately normally distributed (bell curve)

```python
import pandas as pd

from sklearn.preprocessing import StandardScaler

df = pd.read_csv("employee_salary.csv")

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print(scaled_df)
```

2. MinMaxScaler

*Avoid When-->Dataset contains many outliers

When to Use-->Need values in a fixed range (0–1)

```python
import pandas as pd

from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv("employee_salary.csv")

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print(scaled_df)
```

3. RobustScaler

When to Use--->Dataset has many outliers

Avoid When--->Data has no significant outliers

```python
import pandas as pd

from sklearn.preprocessing import RobustScaler

df = pd.read_csv("employee_salary.csv")

scaler = RobustScaler()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print(scaled_df)
```

4. Normalizer

When to Use--->Compare the direction/pattern of rows rather than their magnitude

*Avoid When-->Most tabular datasets

```python
import pandas as pd

from sklearn.preprocessing import Normalizer

df = pd.read_csv("employee_salary.csv")

scaler = Normalizer()

scaled_data = scaler.fit_transform(df)

scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

print(scaled_df)
```
