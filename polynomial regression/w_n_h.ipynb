```python
from sklearn.model_selection import train_test_split
```

```python
import pandas as pd
df=pd.read_csv("Weight-Height Polynomial Dataset.csv")
df.info()
```

```python
df.corr()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['Weight'],df['Height'])
```

This indicates a non-linear (curvilinear) relationship between the variables.

Since your data is curved, linear regression may not capture the pattern perfectly.

```python
from sklearn.linear_model import LinearRegression
```

```python
x=df[['Height']]
y=df['Weight']
xtrain,xtest , ytrain , ytest = train_test_split (x,y,train_size = 0.8 , random_state=43)
model=LinearRegression()
model.fit(xtrain,ytrain)         
```

```python
ypred=model.predict(xtrain)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error,r2_score
```

```python
score = r2_score (ytrain,ypred)
print(score)
```

```python
#from r2 score (0.9352280016563758) we can conclude that model captures patterns well
```

```python
rmse = root_mean_squared_error(ytrain , ypred)
print(rmse)
```

```python
mae = mean_absolute_error(ytrain,ypred)
print(mae)
```

The scatter plot indicates a non-linear relationship between the variables. Although the Linear Regression model achieved an R² score of 0.9352 with RMSE of 6.13 and MAE of 5.24, it may not fully capture the underlying pattern of the data

```python
test_mae = mean_absolute_error(ytest,test_pred)
print(test_mae)
```

The dataset is non-linear, so linear regression gives reasonably good results but is not the most suitable model.

The Linear Regression model achieved a test R² score of 84.1%, indicating that it explains a significant portion of the variance in the data. The RMSE of 6.20 and MAE of 5.32 indicate moderate prediction errors. Since the dataset exhibits a non-linear relationship, the linear model provides acceptable but not optimal performance. A Polynomial Regression model would likely capture the non-linear pattern more effectively and improve prediction accuracy.
