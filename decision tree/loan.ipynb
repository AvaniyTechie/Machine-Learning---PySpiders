```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor

from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn.metrics import  recall_score,precision_score
from sklearn.metrics import f1_score,classification_report
```

```python
df=pd.read_csv("loan_approval_dataset.csv")
df.head()
```

```python
x=df.drop(columns='Loan_Approved')
y=df.Loan_Approved
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,random_state=42,train_size=0.7)
```

```python
print(xtrain["Loan_Term"].unique())
print(xtrain["Employment_Status"].unique())
print(xtrain["Residence_Type"].unique())
print(xtrain["Previous_Default"].unique())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder

cat_cols=["Loan_Term","Employment_Status","Residence_Type","Previous_Default"]
preprocessor=ColumnTransformer(
    transformers=[
        ("categorical_col",OrdinalEncoder(),cat_cols)],
        remainder="passthrough" )  # Keep numeric columns
```

```python
from sklearn.pipeline import Pipeline
model1=Pipeline([
    ("preprocessor",preprocessor),
    ("classifier",DecisionTreeClassifier(random_state=42))
])
```

```python
model1.fit(xtrain,ytrain)
```

```python
ypred_train=model1.predict(xtrain)
ypred_test = model1.predict(xtest)
```

```python
print("Confusion Matrix")
print(confusion_matrix(ytrain, ypred_train))

print("Precision :", precision_score(ytrain, ypred_train,pos_label="Yes"))

print("Recall :", recall_score(ytrain, ypred_train,pos_label="Yes"))

print("Accuracy :", accuracy_score(ytrain, ypred_train))

print("F1 Score :", f1_score(ytrain, ypred_train,pos_label="Yes"))

print(classification_report(ytrain, ypred_train))
```
