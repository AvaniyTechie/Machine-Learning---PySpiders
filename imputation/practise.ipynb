```python
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = {
    'student id': [1,2,None,4,5,6, 7,None , 9 , 10],
    'gender': ['M',None,'F','M',None, 'F','M',None,'M','F'],
    'AGE' : [20,21,None,22,None,24,None,27,21,24],
    'surname': ['aa','bb',None,'cc','dd',None,'ee',None,'ab','av']}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
imputer = SimpleImputer(strategy="constant",fill_value='0')
imputer.fit_transform(df)
```

```python
imputer = SimpleImputer(strategy='constant',fill_value='unknown')
imputer.fit_transform(df[['gender','surname']])
```
