```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler
import seaborn as sns
import numpy as np
from xgboost import XGBClassifier,XGBRegressor
```

```python
df=pd.read_csv("loan_approval_dataset.csv")
df.head()
```

```python
df.info()
```

```python
df['Loan_Approved']=df['Loan_Approved'].replace('Yes',1)
df['Loan_Approved']=df['Loan_Approved'].replace('No',0)
df
```

```python
x = df.drop("Loan_Approved", axis=1)

print(x)

y = df["Loan_Approved"]
print(y)
```

```python
num_cols = x.select_dtypes(include='number').columns
obj_cols = x.select_dtypes(include='object').columns
```

```python
x[obj_cols]=x[obj_cols].astype('category')
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    x,y,
    random_state=42,
    train_size=0.8
)
```

```python
model=XGBClassifier(enable_categorical=True)

# xgb classifer and categorical both TARGET column values are numerical
# handle only feature categorical column
# n_estimators=100,max_depth=None,learning_rate=0.1
```

```python
model.fit(xtrain,ytrain)
```

```python
model.predict(xtrain)
```
