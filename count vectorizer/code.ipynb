COUNT VECTORIZER

```python
from sklearn.feature_extraction.text import CountVectorizer

import pandas as pd
from sklearn.model_selection import train_test_split,GridSearchCV
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("IMDB Dataset.csv")
df.head()
```

```python
x = df["review"]
# x = df[["review"]]    # ❌ 2D DataFrame

y = df["sentiment"]       #target should be 1 col
```

```python
print("x Shape :", x.shape)
print("y Shape :", y.shape)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(
    x,y,
    test_size=0.2,
    random_state=42
)
```

```python
pipe = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("model", LogisticRegression())
])
```

```python
# Hyperparameters
param_grid = {
    "vectorizer__max_features": [1000, 3000, 5000],
    "vectorizer__ngram_range": [(1,1), (1,2)],
    "model__C": [0.1, 1, 10],
    "model__solver": ["liblinear", "lbfgs"]
}
```

```python
# Grid Search
grid = GridSearchCV(
    estimator=pipe,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
grid.fit(xtrain, ytrain)
```

```python
print("Best Parameters:")
print(grid.best_params_)
```

```python
print("\nBest CV Accuracy:")
print(grid.best_score_)
```

```python
# Prediction
y_pred = grid.predict(xtest)
```

```python
print("\nTest Accuracy:")
print(accuracy_score(ytest, y_pred))
```

```python
cv = grid.best_estimator_.named_steps["vectorizer"]
```

```python
print("\nBefore Transformation")
print(type(xtrain))
print("Shape:", xtrain.shape)
```

```python
# Transform
X_train_cv = cv.transform(xtrain)
```

```python
print("\nAfter Transformation")
print(type(X_train_cv))
print("Shape:", X_train_cv.shape)
```

```python
X_array = X_train_cv.toarray()
```

```python
print("\nArray Shape:")
print(X_array.shape)
```

```python
print("\nFirst 5 Rows:")
print(X_array[:5])
```

```python
feature_names = cv.get_feature_names_out()

print("\nFeature Names:")
print(feature_names)
```

```python
print("\nFirst 20 Feature Names:")
print(feature_names[:20])
```

```python
updated_xtrain = pd.DataFrame(
    X_array,
    columns=feature_names
)
```
